```python
team_counts = df['Team'].value_counts()
team_percentages = df['Team'].value_counts(normalize=True) * 100
print(team_counts)
print(team_percentages)
```

```python
team_distribution = pd.DataFrame({
    'Employee Count': team_counts,
    'Percentage': team_percentages.round(2)
})
team_distribution
```

```python
position_counts = df['Position'].value_counts()

# Display the counts
print(position_counts)
```

```python
age_bins = [10,20, 30, 40, 50, 60, 70] 
age_labels = ['10-20','20–30', '31–40', '41–50', '51–60', '61+']
df['age_group'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels, include_lowest=True)
df[['Age', 'age_group']].head()
```

```python
age_group_counts = df['age_group'].value_counts().sort_index()
print(age_group_counts)
```

```python
print("The predominant age group is:" ,age_group_counts.idxmax())
```

```python
salary_by_team_pos = df.groupby(['Team', 'Position'])['Salary'].sum().reset_index()
print(salary_by_team_pos)
```

```python
salary_by_team_pos_sorted=salary_by_team_pos.sort_values(by='Salary', ascending=False)
salary_by_team_pos_sorted.head()
```

```python
correlation = df['Age'].corr(df['Salary'])
print("Correlation between age and salary: ",correlation)
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Age', y='Salary', data=df, alpha=0.6)
sns.regplot(x='Age', y='Salary', data=df, scatter=False, color='red')  # regression line
plt.title("Correlation Between Age and Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.tight_layout()
plt.show()
```

its positively correlated

```python
team_counts = df['Team'].value_counts()
team_percent = (team_counts / len(df)) * 100

# Bar chart
plt.figure(figsize=(12,8))
sns.barplot(x=team_counts.index, y=team_counts.values, hue=team_counts.index, dodge=False, palette='Set2', legend=False)

plt.title("Number of Employees per Team")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='Position', order=df['Position'].value_counts().index, hue='Position', palette='Set3')
plt.title("Employee Count by Position")
plt.xlabel("Position")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=age_group_counts.index, y=age_group_counts.values,hue=age_group_counts, palette='coolwarm')
plt.title("Age Group Distribution of Employees")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
top10_salary_expenditure = salary_by_team_pos_sorted.head(10)
plt.figure(figsize=(12, 6))
sns.barplot(data=top10_salary_expenditure, x='Team', y='Salary', hue='Position')
plt.title("Top 10 Salary Expenditures by Team and Position")
plt.xlabel("Team")
plt.ylabel("Total Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Age', y='Salary', data=df, alpha=0.6)
sns.regplot(x='Age', y='Salary', data=df, scatter=False, color='red')  # regression line
plt.title("Correlation Between Age and Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.tight_layout()
plt.show()
```
